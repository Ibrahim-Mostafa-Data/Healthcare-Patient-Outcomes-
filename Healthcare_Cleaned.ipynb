```python
# Categorical Consistency: Outliers
Q1=df['Billing Amount'].quantile(0.25)
Q2=df['Billing Amount'].quantile(0.50)
Q3=df['Billing Amount'].quantile(0.75)

IQR = Q3 - Q1

Lower_Bound = Q1 - 1.5 *IQR
Upper_Bound = Q3 + 1.5 *IQR
```

```python
IQR
```

```python
outliers = df[
    (df["Billing Amount"] < Lower_Bound) |
    (df["Billing Amount"] > Upper_Bound)
]

outliers
```

```python
# Remove invalid negative Billing Amounts
df = df[df["Billing Amount"] >= 0]

# Validation
df[df["Billing Amount"] < 0]
```

```python
# Another way to remove invalid negative Billing Amounts
df.drop(df[df["Billing Amount"] < 0].index, inplace=True)
```

```python
df.shape
```

```python
df['Admission Year']= df['Date of Admission'].dt.year
df['Admission Month']= df['Date of Admission'].dt.month
df['Length of Stay'] =(df['Discharge Date']-df['Date of Admission']).dt.days
```
